# First load all test images for the test case `a0_300dpi_double_4xlong_2xshort`

```python

from pathlib import Path

# noinspection unused-imports
import IPython.display as pyd
# noinspection unused-imports
import cv2 as cv2
# noinspection unused-imports
import ipympl
# noinspection unused-imports
import ipywidgets as widgets
# noinspection unused-imports
import matplotlib as mpl
# noinspection unused-imports
import matplotlib.pyplot as plt

%matplotlib widget

mpl.rcParams['figure.figsize'] = (10, 10)

type Cv2Img = cv2.typing.MatLike

test_case = "a0_300dpi_double_4xlong_2xshort"

test_data_dir = Path(r"../../test-data")
test_data_input_dir = test_data_dir / "input"
test_data_output_dir = test_data_dir / "output"
test_case_input_dir = test_data_input_dir / test_case
test_case_output_file = test_data_output_dir / f"{test_case}.png"

test_images_paths_with_images: list[tuple[Path, Cv2Img]] = [
    (testImagePath, test_image)
    for testImagePath in test_case_input_dir.iterdir()
    if (test_image := cv2.imread(str(testImagePath))) is not None
]

test_images_paths_with_images = [
    (testImagePath, test_image)
    for testImagePath, test_image
    in test_images_paths_with_images
    if "fwd" in testImagePath.name
]

test_image_2_path: dict[int, Path] = {id(testImage): testPath for testPath, testImage in test_images_paths_with_images}

test_images: list[cv2.typing.MatLike] = [test_image for _, test_image in test_images_paths_with_images]


def test_image_path(test_image: Cv2Img) -> Path:
    return test_image_2_path[id(test_image)]


def test_image_by_idx(test_image_idx: int) -> Cv2Img:
    return test_images[test_image_idx]


print(f"Loaded {len(test_images)} images from {test_case_input_dir}")
```

# Output all images that will be used with some additional data

```python
image_outputs = []

for test_image in test_images:
    image_output = widgets.Output(
        layout=widgets.Layout(
            width="fit-content",
            flex="0 0 auto",
            margin="0 12px 12px 0",
        )
    )
    image_outputs.append(image_output)
    with image_output:
        h, w = test_image.shape[:2]

        test_image_name = test_image_path(test_image).name
        test_image_title = f"{test_image_name} ({w} x {h})"
        print(f"{test_image_title}")

        if w < h:
            #cv2.imshow(test_image_title, test_image)
            plt.imshow(cv2.cvtColor(test_image, cv2.COLOR_BGR2RGB))
            plt.axis("off")
            plt.show()
        else:
            #cv2.imshow(test_image_title, test_image)
            plt.imshow(cv2.cvtColor(test_image, cv2.COLOR_BGR2RGB))
            plt.axis("off")
            plt.show()

display(widgets.Box(
    children=image_outputs,
    layout=widgets.Layout(
        display="flex",
        flex_flow="row wrap",
        align_items="flex-start",
        width="100%",
    ),
))
```
